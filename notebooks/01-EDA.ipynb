# EDA: Superstore Sales Insights
Starter notebook for exploratory data analysis. Run cells step-by-step in Codespaces.

This notebook will attempt to run data/download_data.py to fetch the Excel file into data/ if it is not present.

```python
# Basic imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

# Ensure data directory exists and download dataset if missing
import pathlib, subprocess, sys
data_path = pathlib.Path('data/Global_Superstore.xlsx')
if not data_path.exists():
    print('Dataset not found at', data_path)
    print('Attempting to run data/download_data.py...')
    try:
        # run the helper script from the repository root
        subprocess.run([sys.executable, 'data/download_data.py'], check=True)
    except Exception as e:
        print('Download helper failed:', e)

print('Dataset exists:', data_path.exists())
```

```python
# Load the dataset (uncomment when the file is present)
import os
if os.path.exists('data/Global_Superstore.xlsx'):
    df = pd.read_excel('data/Global_Superstore.xlsx', engine='openpyxl')
    print('rows, cols:', df.shape)
    display(df.head())
else:
    print('Data file not found. Run the download helper or place file at data/Global_Superstore.xlsx')
```
